#### 직접 호출하기

```python
import requests
```

```python
info = {
    # 사용할 모델
    'model' : "qwen3:0.6b",
    'prompt' : "너는 여행 전문가야. 한국의 여행지 중에 아름다운 곳 3가지를 추천해줘. 모든 응답은 한글로 작성해줘",
    # invoke처럼 응답을 한번에 JSON으로 받음
    'stream' : False
}
```

```python
try:
    response = requests.post('http://localhost:11434/api/generate', json=info, timeout=120)
    # 응답코드가 정상(200)이 아니면 예외를 발생
    response.raise_for_status()
    # 응답 JSON 가져오기
    data = response.json()
    # 응답 출력
    print(data.get("response"))
except Exception as e:
    print(e)
```

## Ollama_LangChain 구성하기

```python
!pip install langchain langchain-ollama langchain-core
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
```

```python
# 모델생성
llm = ChatOllama(
    model = "qwen3:0.6b",
    temperature = 0.7
)
```

```python
# 프롬프트
template = ChatPromptTemplate.from_messages([
    ('system', '당신은 한국여행 전문가입니다. 아래요청사항에 대해 한국어와 한국지역기준으로 답변해주세요'),
    ('human', 'request : {input}')
])

# 체인구성
chain = template | llm | StrOutputParser()
```

```python
# 체인호출
res = chain.invoke({"input" : "한국에서 여행하기 좋은 바닷가 알려줘."})
```

```python
print(res)
```
