```python
# temperature : 0-1 -> 1에 가까울 수록 자유도가 높음
# max_tokens : 사용할 수 있는 최대 토큰 (입력토큰(질문) + 응답토큰 (AI의 응답))
llm = init_chat_model(model="gpt-4o-mini",
                      temperature=0,
                      max_tokens=500)

# 변수가 포함된 프롬프트
template = "{country}의 수도는 어디야 ?"

# PromptTemplate : 템플릿 -> 프롬프트 변환
# invoke() 사용하기 전에 변수값을 할당해서 프롬프트롤 생성한 후에 모델에 전달
# 방법 1
prompt1 = PromptTemplate(template=template, input_variables=["country"])

# 방법 2
prompt2 = PromptTemplate.from_template(template)
```

```python
chain = prompt2 | llm

# (1) 변수값을 템플리의 변수에 할당해서 프롬프트를 완성
# (2) 완성된 프롬프트를 LLM에 전달
# (3) LLM이 응답 생성해서 반환
answer = chain.invoke({"country":"중국"})
print(answer)
print(answer.content)
```

- 구조화된 프롬프트 : 단계별 지시사항

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 코드 리뷰 전문가입니다."),
    ("human", """다음 코드를 리뷰해주세요:

<code>
{code}
</code>

다음 순서로 분석해주세요:
1. **기능 요약**: 코드가 수행하는 작업 설명
2. **장점**: 잘 작성된 부분 (2-3가지)
3. **개선점**: 수정이 필요한 부분 (2-3가지)
4. **리팩토링 제안**: 개선된 코드 예시

각 섹션은 명확히 구분하여 작성해주세요.""")
])

chain = prompt | llm

code = """
def sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# 테스트
data = [64, 34, 25, 12, 22, 11, 90]
print("처리 전", data)
print("처리 후", sort(data))
"""

answer = chain.invoke({"code": code})

display(Markdown(answer.content))
```

# 체인 실행 명령 : batch()

- batch() : 입력 리스트에 대해 chain을 호출하는 함수

```python
lim = init_chat_model("gpt-4o-mini")

prompt = ChatPromptTemplate.from_template("{country}의 수도는 어디야")

chain = prompt | llm

country_list = [
    {"country":"대한민국"},
    {"country":"중국"},
    {"country":"일본"},
    {"country":"프랑스"},
    {"country":"영국"},
]

answers = chain.batch(country_list)

for answer in answers:
    display(Markdown(answer.content))
```

- stream() : 스트림 출력

```python
llm = init_chat_model("gpt-5.4")

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "너는 여행 및 음식 전문가이다"),
        ("human", "{input}")
    ]

)

chain = prompt | llm

for answer in llm.stream("광주의 유명한 빵집 3군데를 추천해줘 (대표 제품도 소개해줘)"):
        print(answer.content, end="")
```

# 비동기 함수 사용하기 (선택)

- <font color=red>astream()</font> : 비동기적으로 응답의 청크를 스트리밍
- <font color=red>abatch()</font> : 비동기적으로 입력 목록에 대해 체인을 호출

- ainvoke()

- RunnableBranch()
  - 조건문과 RunnableLambda()를 사용한 라우팅을 더 활용

```python
# 한국어, 영어로 질문하는 경우 각각에 맞는 프롬프트를 선택하도록 하는 기능
llm = init_chat_model("gpt-4o-mini")

k_prompt = ChatPromptTemplate.from_template(
    "한국어로 된 다음과 같은 질문에 대해 한국어로 대답해줘 : {question}"
)

e_prompt = ChatPromptTemplate.from_template(
    "Answer the following question in English : {question}"
)

d_prompt = ChatPromptTemplate.from_template(
    "Please Answer : {question}"
)

#질문에 대한 언어 인식 함수
def detect_language(input_dict) :
    # 딕셔너리에서 question 키 값을 가져온다 (사용자 질문)
    question = input_dict.get("question", "")

    # "\uac00" ~ "\ud7a3" : 한글에 해당하는 코드의 범위 -> 한글이라면
    if any("\uac00" <= char <= "\ud7a3" for char in question):
        return "korean"
    return "english"

#3가지 경우의 실행 분기문 설정 (한국어, 영어, 기타)
chain = RunnableBranch(
    (lambda x : detect_language(x) == "korean", k_prompt | llm),
    (lambda x : detect_language(x) == "english", e_prompt | llm),
    d_prompt | llm
)

for answer in llm.stream("인공지능이란"):
    print(answer.content, end="")


answer = chain.invoke({"question" : "AI is?"})
display(Markdown(answer.content))
```

- with_structured_output
  - LangChain 1.0에서 권장하는 구조화 출력 방식
  - 모델의 네이티브 JSON 스키마 기능을 활용
  - LLM 모델에 설정 구조(class)로 출력하는 기능을 붙여주는 기능

```python
#구조화된 형식 정의
#Field : 변수에 대한 설명이나 조건 등을 설정

class MovieReview(BaseModel) :
    """영화 리뷰 정보"""
    title : str = Field(description="영화 제목")
    rating : int = Field(description="영화 평점 (1-10점)")
    pros : str  = Field(description="영화 장점")
    cons : str  = Field(description="영화 단점")
    summary : str  = Field(description="한 줄 요약")
```

```python
# 기존 LLM에 구조화된 출력을 만드는 기능을 추가
strudcted_llm = llm.with_structured_output(MovieReview)

answer = strudcted_llm.invoke("군체 영화에 대해 설명해줘")

print(f"제목 : {answer.title}")
print(f"평점 : {answer.rating}")
print(f"장점 : {answer.pros}")
print(f"단점 : {answer.cons}")
print(f"요약 : {answer.summary}")
```

- 실습 : 코드 리뷰 기능 구현
    - 코드를 입력받아서 해당 코드에 대해 분석하는 기능
    - 구조화된 데이터
        - 코드 기능 요약 (summary)
        - 1-10점 품질 평가 (score)
        - 문제점 (issues)
        - 개선할 내용 (suggestions)



```python
class CodeReview(BaseModel) :
    """코드 리뷰 정보"""
    summary : str = Field(description="코드 기능 요약")
    score : int = Field(description="1-10점 품질 평가")
    issues : str  = Field(description="문제점")
    suggestions : str  = Field(description="개선할 내용")
```

```python
llm = init_chat_model("gpt-5.4")

strudcted_llm = llm.with_structured_output(CodeReview)

input_code = input(""" code : """)

answer = strudcted_llm.invoke(input_code)

print(f"요약 : {answer.summary}")
print(f"평점 : {answer.score}")
print(f"문제점 : {answer.issues}")
print(f"개선할 내용 : {answer.suggestions}")
```
